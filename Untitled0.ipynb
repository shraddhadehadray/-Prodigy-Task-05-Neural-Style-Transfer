```python
# TASK 05 - Neural Style Transfer

!pip install -q tensorflow tensorflow-hub matplotlib pillow

import tensorflow as tf
import tensorflow_hub as hub
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image
from google.colab import files

# Upload images
print("Upload CONTENT IMAGE (landscape/photo):")
content_upload = files.upload()
content_path = list(content_upload.keys())[0]

print("Upload STYLE IMAGE (painting):")
style_upload = files.upload()
style_path = list(style_upload.keys())[0]


# Load images
def load_image(path):
    img = Image.open(path).convert("RGB")
    img = img.resize((256, 256))
    img = np.array(img).astype(np.float32) / 255.0
    return img[np.newaxis, ...]

content_image = load_image(content_path)
style_image = load_image(style_path)

# Load pre-trained Neural Style Transfer model
model = hub.load(
    "https://tfhub.dev/google/magenta/arbitrary-image-stylization-v1-256/2"
)

# Apply style transfer
stylized_image = model(
    tf.constant(content_image),
    tf.constant(style_image)
)[0]

# Display result
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(content_image[0])
plt.title("Content Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(style_image[0])
plt.title("Style Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(stylized_image[0])
plt.title("Neural Style Transfer Output")
plt.axis("off")

plt.show()

# Save output
output = np.clip(stylized_image[0].numpy() * 255, 0, 255).astype(np.uint8)
Image.fromarray(output).save("task05_neural_style_transfer.jpg")

print("Task 05 completed successfully!")
```
